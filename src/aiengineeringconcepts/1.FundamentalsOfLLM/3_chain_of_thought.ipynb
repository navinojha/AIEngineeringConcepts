This Excersie focuses on Chain Of Thought Limitations for LLM's using GSM8K Dataset, which Grade school mathematics quality > 8500, where the base model fails to reason and provide
the correct answer

```python
import os

from groq import Groq
from dotenv import load_dotenv

load_dotenv()

Q = ("Navin has 5 balls, He buys 3 more cans and each can has 4 balls ",
     "so how many total balls will Navin have ?")

client = Groq()

# Demo: For failure of LLM's 
fail = client.chat.completions.create(
    model="qwen/qwen3.8-27b",
    messages=[
        {
            "role": "system",
            "content": "The answer (Just the number) is : "
        },
        {
            "role": "user",
            "content": "Navin has 5 balls, He buys 3 more cans and each can has 4 balls, so how many total balls will Navin have ?"
        }
    ],
    max_completion_tokens=20,
    include_reasoning=False
)

print("Groq Model says: ", fail.choices[0].message.content)
```
